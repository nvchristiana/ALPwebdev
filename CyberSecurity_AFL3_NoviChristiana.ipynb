```python
import hashlib

def md5_hash(text):
    return hashlib.md5(text.encode()).hexdigest()
```

```python
target_a = "0110b96270248f746ecca06f1ce09746"

for i in range(1000000):
    pin = f"{i:06d}"
    if md5_hash(pin) == target_a:
        print("PIN Target A:", pin)
        break
```

```python
target_b = "8d4c531eb4b0f54df72aa6839abbeaec"

found_b = False
for salt in range(10):
    for i in range(1000000):
        pin = f"{i:06d}"
        if md5_hash(f"{salt}{pin}") == target_b:
            print("PIN Target B:", pin)
            print("Salt Target B:", salt)
            found_b = True
            break
    if found_b:
        break
```
